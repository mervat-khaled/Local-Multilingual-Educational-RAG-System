#  Human Evaluation

***Multi-Criteria Scoring:** Human raters grade output pairs on a strict 0–3 scale across 5 dimensions: *Relevance, Faithfulness, Latency, Readability,* and *Perplexity*.

### Human Evaluation Guidelines:

#### 1- Relevance
Definition: Measures how well the AI’s answer addresses the specific question asked by the student. It penalizes answers that are off-topic, incomplete, or dodge the question.

	•	Score 3 (Perfect): The AI directly and completely answers the student’s question. It provides exactly what was asked without unnecessary filler.
	•	Score 2 (Partial/Good): The AI addresses the main topic but misses a minor part of the student's question, or includes slightly irrelevant extra information.
	•	Score 1 (Poor): The AI talks about the general topic but fails to answer the specific question asked (e.g., the student asks "How" and the AI answers "When").
	•	Score 0 (Irrelevant): The answer is completely off-topic, talks about something else entirely, or says "I don't know" when the information is basic.


#### 2- Faithfulness (Truthfulness / Groundedness)
Definition: Measures if the AI's answer is strictly based on the provided PDF document chunks. It checks for hallucinations (making things up). Note: An answer can be relevant but unfaithful if the AI makes up a fake fact to answer the question.

	•	Score 3 (Completely Faithful): Every single fact, date, or claim in the AI's response is directly supported by the provided document chunks. No outside knowledge or assumptions are added.
	•	Score 2 (Mostly Faithful): The core answer is fully supported by the document, but the AI adds a small, harmless piece of outside information or common knowledge that wasn't explicitly in the PDF.
	•	Score 1 (Low Faithfulness): The AI introduces significant information, equations, or facts that cannot be found anywhere in the provided document, or mildly distorts what the document says.
	•	Score 0 (Hallucination): The AI completely fabricates information, contradicts the source document, or claims the document says something it flatly does not.
    

#### -3 Latency (Perceived Speed)
Definition: Measures how the generation speed impacts the student's user experience. Grade this based on your perceived wait time during testing.

	•	Score 3 (Instant): The response appears almost immediately (under 2 seconds) or streams so smoothly that there is zero disruption to a student's train of thought.
	•	Score 2 (Acceptable Delay): There is a brief, noticeable pause (3–5 seconds) before the answer appears. It is slightly annoying but acceptable for an educational tool.
	•	Score 1 (Frustratingly Slow): The delay is significant (6–12 seconds). A student would likely lose focus, get distracted, or assume the app is lagging.
	•	Score 0 (Unusable / Timeout): The system takes longer than 12 seconds to respond, throws a loading error, or forces you to refresh the page.
    

#### 4- Readability (Formatting & Structure)
Definition: Measures how well the information is structured for a student. It evaluates formatting, visual layout, and pedagogical clarity (e.g., use of bullet points, bold text, and proper paragraph breaks).

	•	Score 3 (Excellent Structure): Highly scannable. It uses bullet points, numbered lists, or bold text appropriately to highlight key educational concepts. It looks clean and easy to study.
	•	Score 2 (Average Structure): The response is organized into readable paragraphs, but it lacks visual anchors like bullet points or bold text that would make it easier for a student to quickly digest.
	•	Score 1 (Poor Structure): The response is a massive, daunting "wall of text" with no paragraph breaks, or it contains messy layout artifacts (like random spaces or broken characters) from the PDF conversion.
	•	Score 0 (Unreadable): The formatting is completely broken. Sentences are randomly cut off, text overlaps, or the layout makes it impossible to read comfortably.

#### 5- Perplexity (Human Fluency & Tone)
Definition: In human evaluation, this measures how natural, grammatically correct, and cohesive the language is. It evaluates if the AI sounds like a professional human teacher or a broken machine.

	•	Score 3 (Natural/Human-like): The language is flawless. It reads like it was written by a human educator. Perfect grammar, smooth transitions, and an encouraging, professional tone.
	•	Score 2 (Slightly Robotic): The text is grammatically correct and fully understandable, but the phrasing feels a bit robotic, clinical, or repetitive.
	•	Score 1 (Awkward / Clunky): The phrasing is highly unnatural. Thoughts are disjointed, sentences are clunky, or there are clear grammatical errors that force you to reread the sentence to understand it.
	•	Score 0 (Gibberish): Complete word salad. The sentences make no linguistic sense, words are used completely out of context, or the text is a repetitive loop.

### Evaluation Golden Rules for Raters:

	1.	Keep criteria independent: An AI can give a beautifully formatted (Readability = 3) and perfectly fluent (Perplexity = 3) answer, but make up a complete lie (Faithfulness = 0). Grade each category strictly on its own definition.
	2.	Read the Source Document Chunks First: Do not grade Faithfulness based on what you know to be true; grade it only on whether that information exists in the specific PDF chunks provided for that query.

 

# Understanding Inter-Annotator Agreement (Cohen's Kappa)

### 1. What is Cohen's Kappa (κ)?
Cohen’s Kappa is a statistical metric used to measure agreement between exactly two raters who are grading items into distinct categories (like our 0, 1, 2, 3 scores).
Unlike simple percentage agreement (e.g., "They agreed 80% of the time"), Cohen's Kappa subtracts the probability of accidental agreement. If both raters blindly guessed "3" all day, their percentage agreement would be high, but their Kappa score would be 0.


### 2. How to Interpret the Kappa Score (κ)
The resulting score falls between -1 and 1, though in practice it is usually between 0 and 1:

	•	0.81 – 1.00: Almost Perfect Agreement. Your guidelines are crystal clear, and your raters are completely aligned.
	•	0.61 – 0.80: Substantial Agreement. This is the gold standard target for AI evaluation datasets. The data is highly reliable.
	•	0.41 – 0.60: Moderate Agreement. Acceptable, but indicates that some metrics (likely Perplexity or Readability) are too subjective. You need to clarify your guidelines.
	•	Below 0.40: Fair to Poor Agreement. The raters are contradicting each other. Your evaluation data cannot be trusted. You must retrain the raters.

```python
all_evaluaterA=pd.concat([evaluater_A,evaluater_A2])
all_evaluaterB=pd.concat([evaluater_B,evaluater_B2])
```

```python
all_evaluaterA.info()
```

```python
columns_to_evaluate = [' Relevance', 'Faithfulness', ' Latency', 'Readability', 'Preplexity']
```

```python
all_evaluater_a_labels = np.concatenate([all_evaluaterA[col].values for col in columns_to_evaluate])
all_evaluater_b_labels = np.concatenate([all_evaluaterB[col].values for col in columns_to_evaluate])

overall_kappa = cohen_kappa_score(all_evaluater_a_labels, all_evaluater_b_labels,weights='linear')

print(f"Overall Cohen's Kappa Score: {overall_kappa:.4f}")
if overall_kappa > 0.80:
    print("Interpretation: Almost Perfect Agreement! Your guidelines are rock solid.")
elif overall_kappa > 0.60:
    print("Interpretation: Substantial Agreement. Highly reliable evaluation data.")
elif overall_kappa > 0.40:
    print("Interpretation: Moderate Agreement. Consider clarifying borderline rubric rules.")
else:
    print("Interpretation: Poor Agreement. Raters are misaligned. Stop and retrain them.")
```
