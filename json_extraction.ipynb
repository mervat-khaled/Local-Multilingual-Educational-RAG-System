```python
import json
import os

directory_path = '.'
data =None

for filename in os.listdir(directory_path):
    if filename.endswith('.json'):
        file_path = os.path.join(directory_path, filename)
        
        with open(file_path, 'r', encoding='utf-8') as file:
            data = json.load(file)
            
            print(f"Successfully loaded: {filename}")
```

```python
chat_df= pd.json_normalize(data)
```

```python
chat_df.head(2)
```

```python
chat_df = chat_df.rename(columns={
    'metrics.execution_time_seconds': 'execution_time_seconds',
    'metrics.retrieved_context': 'retrieved_context'
})
```

```python
chat_df=chat_df[['question','answer','execution_time_seconds','retrieved_context']]
```

```python
chat_df.info()
```
